```python
def reversenum(n, rev=0):
    if n == 0:
        return rev
    digit = n % 10
    return reversenum(n // 10, rev * 10 + digit)

n = 101
print(reversenum(n))
```

```python
arr=[5,6,2,3,4,9,4]
arr=sorted(arr)
left=0
right=len(arr)-1
target=8
while left<right:
  ele=arr[left]+arr[right]
  if ele==target:
    print(arr[left],arr[right],"=",target)
    break
  elif ele<target:
    left=left+1
  else:
    right=right-1
```
