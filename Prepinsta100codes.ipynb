```python
class Solution:
    def plusOne(self, digits: list[int]) -> list[int]:
        l1 = len(digits)
        newlist = []

        for i in range(l1):
            if i < l1 - 1:
                newlist.append(digits[i])
            else:
                newlist.append(digits[i] + 1)

        return newlist

# Create an object of the class
obj = Solution()

# Call the function with a list
result = obj.plusOne([1, 2, 3])
print(result)
```
