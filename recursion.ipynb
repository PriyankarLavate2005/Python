```python
def print_hello(n):
    if n <= 0:
        return
    print("Hello, World!")
    print_hello(n - 1)

# Call the function to print it 5 times
print_hello(5)


```

```python

def add(n):
  count=0
  if n<0:
    return
  for i in range(0,n):
    count=count+i
    print(count)
  add(n-1)
add(6)
```

```python
def printnum(n):
    if n<1:
      return
    print(n)
    printnum(n-1)
printnum(5)
```

```python
def factorial(n):
  if n==1 or n==0:
    return 1
  return n*factorial(n-1)
print(factorial(3))

```

```python
def firstNnaturalnumbers(n):
  if n==0:
    return 0
  return n+firstNnaturalnumbers(n-1)
print(firstNnaturalnumbers(12))
```

```python
def powerof2(n):
    count = 0
    if n <= 0:
        return False, count

    while n % 2 == 0:   # keep dividing by 2
        n = n // 2
        count += 1

    if n == 1:   # means it was a power of 2
        return True, count
    else:
        return False, count

print(powerof2(4))   # (True, 2) -> 4 = 2^2
print(powerof2(6))   # (False, 1) -> stopped early
print(powerof2(8))   # (True, 3) -> 8 = 2^3
```

```python
def printdigits(n):
  if n < 10:
    print(n)
  else:
    printdigits(n // 10)
    print(n % 10)

printdigits(34565)
```
