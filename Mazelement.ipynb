```python
def maxelement(arr):
  n=len(arr)
  max=0
  for i in range(0,n):
    if arr[i]>max:
      max=arr[i]
  print(max)
maxelement([6,7,3,23,8,1,2])
```

```python
def minelement(arr):
  n=len(arr)
  min=arr[0]
  for i in range(0,n):
    if min>=arr[i]:
      min=arr[i]
  print(min)

minelement([6,7,3,23,8,2])
```
