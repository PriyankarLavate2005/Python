```python
def factorial(n):
    # base case
    if n == 0 or n == 1:
        return 1
    # recursive case
    else:
        return n * factorial(n - 1)
# take input
num = int(input("Enter a number: "))
fact=factorial(num)
def newfact(n):
  total=0
  lastdigit=0
  while n>0:
    lastdigit=n%10
    total=total+factorial(lastdigit)
    n=n//10
  if total==num:
    print("strong number")
  else:
    print("not strong number")
newfact(num)
```
