```python

t = int(input())
for _ in range(t):
    a = int(input())
    b = int(input())
    c = int(input())

    list1 = [a, b, c]          # reset list for each test case

    maxn = list1[0]            # start with first element
    minn = list1[0]


    for i in list1:
        if i > maxn:
            maxn = i
        if i < minn:
            minn = i

    print(maxn)
    print(minn)
```
