```python
def find_max(nums):
  max=nums[0]
  for num in nums:
    if num>max:
      max=num
  print(max)
find_max([4,5,6,78,43])
```

```python
def find_secondLargest(nums):
    if len(nums) < 2:
        print("Not enough elements")
        return

    firstlargest = secondlargest = float('-inf')

    for num in nums:
        if num > firstlargest:
            secondlargest = firstlargest
            firstlargest = num
        elif num > secondlargest and num != firstlargest:
            secondlargest = num

    if secondlargest == float('-inf'):
        print("No valid second largest found")
    else:
        print(secondlargest)

# Run it with your list
find_secondLargest([67, 54, 34, 23, 90])  # Output: 67
```
